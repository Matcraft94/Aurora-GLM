## 4. Mathematical Specification

### Binomial GLM (Logistic Regression)

$$Y_i \sim \text{Bernoulli}(\pi_i)$$

$$\text{logit}(\pi_i) = \log\left(\frac{\pi_i}{1-\pi_i}\right) = \mathbf{x}_i^T \boldsymbol{\beta}$$

### Likelihood and Estimation

The binomial log-likelihood is

$$\ell(\boldsymbol{\beta}) = \sum_{i=1}^n \left[ y_i \log \pi_i + (1 - y_i) \log(1 - \pi_i) \right]$$

maximized by **IRLS** (Iteratively Reweighted Least Squares): each iteration solves a weighted least-squares problem on the working response $z = \eta + (y - \pi)\,\partial\eta/\partial\pi$ with weights $w_i = \pi_i(1-\pi_i)$.

**Deviance** (twice the difference to the saturated model):

$$D = -2 \sum_{i=1}^n \left[ y_i \log \hat{\pi}_i + (1 - y_i) \log(1 - \hat{\pi}_i) \right]$$

Model selection uses AIC $= D + 2p$ and BIC $= D + p\log n$.

### Odds Ratio Interpretation

$$\text{Odds Ratio} = e^{\beta_j}$$

- OR > 1: Increased odds of churn
- OR < 1: Decreased odds of churn

Continuous predictors (tenure, monthly charges) are **standardized**, so their ORs are per standard deviation.

## 5. Model Fitting

````python
# Prepare design matrix
print("="*80)
print("MODEL FITTING")
print("="*80)

# Build feature matrix (no explicit intercept column: fit_glm adds one)
features = []
names = []

if 'tenure_std' in df.columns:
    features.append(df['tenure_std'].values)
    names.append('Tenure')

if 'MonthlyCharges_std' in df.columns:
    features.append(df['MonthlyCharges_std'].values)
    names.append('MonthlyCharges')

for col in ['senior', 'partner', 'dependents', 'paperless']:
    if col in df.columns:
        features.append(df[col].values)
        names.append(col.capitalize())

if 'contract_1yr' in df.columns:
    features.append(df['contract_1yr'].values)
    features.append(df['contract_2yr'].values)
    names.extend(['Contract_1yr', 'Contract_2yr'])

X = np.column_stack(features)
y = df['churn'].values

print(f"Design Matrix: {X.shape}")
print(f"Predictors: {names}")

# Fit model
start_time = time.time()
result = fit_glm(X=X, y=y, family='binomial', link='logit')
time_fit = time.time() - start_time

print(f"\nConverged: {result.converged_}")
print(f"Iterations: {result.n_iter_}")
print(f"Time: {time_fit:.3f}s")
print(f"Intercept: {result.intercept_:.4f}")
print(f"\nAIC: {result.aic_:.2f}")
print(f"BIC: {result.bic_:.2f}")
print("="*80)
````

## 6. Residual Diagnostics

For a binomial GLM, residual checks use **deviance residuals**

$$d_i = \text{sign}(y_i - \hat{\pi}_i)\, \sqrt{-2\left[ y_i \log \hat{\pi}_i + (1-y_i) \log(1-\hat{\pi}_i) \right]}$$

and, because individual 0/1 residuals are uninformative, **binned residuals** (mean residual within groups of similar fitted probability) plus a **calibration plot** (observed churn rate vs mean predicted probability per decile).

````python
print("="*80)
print("RESIDUAL DIAGNOSTICS")
print("="*80)

# Fitted probabilities (use the model's own predict: it adds the intercept)
prob = np.asarray(result.predict(X))
prob_clip = np.clip(prob, 1e-10, 1 - 1e-10)

# Deviance residuals
d = np.sign(y - prob_clip) * np.sqrt(-2 * (y * np.log(prob_clip) + (1 - y) * np.log(1 - prob_clip)))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Panel 1: deviance residuals vs fitted (binned, since 0/1 residuals are uninformative)
bins = pd.qcut(prob, q=20, duplicates='drop')
binned = pd.DataFrame({'prob': prob, 'd': d}).groupby(bins, observed=True).agg(
    mean_prob=('prob', 'mean'), mean_resid=('d', 'mean'))
axes[0].plot(binned['mean_prob'], binned['mean_resid'], 'o-', color='steelblue')
axes[0].axhline(0, color='red', linestyle='--', linewidth=2)
axes[0].set_xlabel('Fitted probability (binned)')
axes[0].set_ylabel('Mean deviance residual')
axes[0].set_title('Binned Deviance Residuals vs Fitted', fontweight='bold')
axes[0].grid(alpha=0.3)

# Panel 2: histogram of deviance residuals
axes[1].hist(d, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
axes[1].axvline(0, color='red', linestyle='--', linewidth=2)
axes[1].set_xlabel('Deviance residual')
axes[1].set_ylabel('Frequency')
axes[1].set_title('Distribution of Deviance Residuals', fontweight='bold')
axes[1].grid(alpha=0.3)

# Panel 3: calibration plot (deciles of predicted probability)
cal_bins = pd.qcut(prob, q=10, duplicates='drop')
cal = pd.DataFrame({'prob': prob, 'y': y}).groupby(cal_bins, observed=True).agg(
    mean_pred=('prob', 'mean'), obs_rate=('y', 'mean'), n=('y', 'size'))
axes[2].plot([0, 1], [0, 1], 'r--', linewidth=2, label='Perfect calibration')
axes[2].scatter(cal['mean_pred'], cal['obs_rate'], s=cal['n'] / 5, alpha=0.7, color='steelblue')
axes[2].plot(cal['mean_pred'], cal['obs_rate'], 'b-', alpha=0.5)
axes[2].set_xlabel('Mean predicted probability')
axes[2].set_ylabel('Observed churn rate')
axes[2].set_title('Calibration Plot (deciles)', fontweight='bold')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\nMean deviance residual: {d.mean():.4f} (should be ≈ 0)")
print(f"SD deviance residual:   {d.std():.4f}")
max_dev = np.abs(binned['mean_resid']).max()
print(f"Max |binned residual|:  {max_dev:.4f} across 20 probability bins")
cal_err = np.abs(cal['obs_rate'] - cal['mean_pred']).mean()
print(f"Mean calibration error: {cal_err:.4f} across 10 deciles")
if cal_err < 0.05:
    print("   ✓ Model is well calibrated (mean error < 5 points)")
else:
    print("   ⚠ Calibration error above 5 points - investigate")
print("\n" + "="*80)
````

## 7. Interpretation and Predictive Performance

Odds ratios on the response scale, then classification metrics (confusion matrix at threshold 0.5) and ROC/AUC.

````python
# Odds ratios
print("="*80)
print("ODDS RATIOS")
print("="*80)

print("\nCoefficients and Odds Ratios:")
print("-" * 60)

print(f"{'Intercept':20s}: coef={result.intercept_:+.4f}")
for name, coef in zip(names, result.coef_):
    odds_ratio = np.exp(coef)
    direction = 'increases' if coef > 0 else 'decreases'
    print(f"{name:20s}: coef={coef:+.4f}, OR={odds_ratio:.3f}")

print("\n" + "="*80)
````

````python
# Predictions and ROC
print("="*80)
print("PREDICTIVE PERFORMANCE")
print("="*80)

# Predicted probabilities (model's own predict: intercept handled internally)
prob = np.asarray(result.predict(X))

# Classification at 0.5 threshold
pred_class = (prob > 0.5).astype(int)
accuracy = (pred_class == y).mean()

# Confusion matrix
tp = ((pred_class == 1) & (y == 1)).sum()
tn = ((pred_class == 0) & (y == 0)).sum()
fp = ((pred_class == 1) & (y == 0)).sum()
fn = ((pred_class == 0) & (y == 1)).sum()

sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nClassification Metrics (threshold=0.5):")
print(f"   Accuracy: {accuracy:.3f}")
print(f"   Sensitivity: {sensitivity:.3f}")
print(f"   Specificity: {specificity:.3f}")
print(f"   Precision: {precision:.3f}")

# ROC curve
thresholds = np.linspace(0, 1, 100)
tpr_list = []
fpr_list = []
for t in thresholds:
    pred = (prob > t).astype(int)
    tpr = ((pred == 1) & (y == 1)).sum() / max(1, (y == 1).sum())
    fpr = ((pred == 1) & (y == 0)).sum() / max(1, (y == 0).sum())
    tpr_list.append(tpr)
    fpr_list.append(fpr)

# AUC (trapezoidal)
auc = -np.trapezoid(tpr_list, fpr_list)
print(f"   AUC: {auc:.3f}")

# Plot ROC
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(fpr_list, tpr_list, 'b-', linewidth=2, label=f'ROC (AUC={auc:.3f})')
ax.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random')
ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curve', fontweight='bold')
ax.legend()
ax.grid(alpha=0.3)
plt.show()

print("="*80)
````

### Statistical Interpretation

**Logistic Regression Equation**:
```
logit(p_churn) = -1.272 - 0.906×tenure_std + 0.748×charges_std 
                 + 0.386×senior + 0.026×partner - 0.240×dependents 
                 + 0.461×paperless - 0.934×contract_1yr - 1.810×contract_2yr
```

**Example Prediction** (high-risk customer):
- Tenure: low (-1 SD) → +0.906
- Monthly charges: high (+1 SD) → +0.748
- Paperless: yes → +0.461
- No contract (baseline) → 0
- **Total logit**: -1.272 + 0.906 + 0.748 + 0.461 = **0.843**
- **Churn probability**: 1/(1+exp(-0.843)) = **69.9%** (high risk)

**Example Prediction** (low-risk customer):
- Tenure: high (+1 SD) → -0.906
- Monthly charges: low (-1 SD) → -0.748
- Contract: 2-year → -1.810
- Has dependents → -0.240
- **Total logit**: -1.272 - 0.906 - 0.748 - 1.810 - 0.240 = **-4.976**
- **Churn probability**: 1/(1+exp(4.976)) = **0.7%** (very low risk)

- Current threshold (0.5) prioritizes overall accuracy
