# Case Study 01: Insurance Pricing with Gamma GLM
## Aurora-GLM Showcase: Gamma Regression for Healthcare Cost Modeling

---

## Overview

This case study demonstrates **Gamma GLM** for modeling medical insurance charges - a classic application of GLMs for positive, right-skewed cost data.

### Research Context

Health insurance companies need to predict medical charges to:
- Set appropriate premiums
- Identify high-risk individuals
- Design targeted wellness programs
- Estimate reserves for claims

**Medical charges** are:
- **Positive**: No zero or negative charges
- **Right-skewed**: Most people have low costs, few have very high costs
- **Heteroscedastic**: Variance increases with mean (high-cost individuals vary more)
- **Multiplicative**: Effects are better expressed as percentages (smokers pay 3× more)

**Dataset**: 1,338 insurance records with demographics and medical charges

### Research Questions

1. **RQ1:** What factors drive medical insurance costs?
2. **RQ2:** How much more do smokers pay compared to non-smokers?
3. **RQ3:** Are there regional differences in healthcare costs?
4. **RQ4:** Which link function is most appropriate (log vs identity)?
5. **RQ5:** Can we identify high-risk profiles for differential pricing?

### Aurora-GLM Capabilities Demonstrated

1. **Gamma family** for positive, right-skewed outcomes
2. **Log link** for multiplicative effects (rate ratios)
3. **Link function comparison** (log vs identity vs inverse)
4. **Model diagnostics** (residuals, Q-Q plots, calibration)
5. **Rate ratio interpretation** (exp(β) for policy communication)
6. **Risk segmentation** (high-risk vs low-risk profiles)
7. **Multi-backend support** (NumPy, PyTorch, JAX)
